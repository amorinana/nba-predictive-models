```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# pd.set_option('display.max_rows', None)
pd.set_option('display.max_columns', None)

df = pd.read_csv('nba_stats.csv')
df.head(3)
```

```python
df['salary'] = (df['salary'].astype(str).str.replace('$', '', regex=False).str.replace(',', '', regex=False).astype(np.float64))
```

```python
df['birthdate_clean'] = df['birthdate'].str.replace(r'\s*\(.*\)', '', regex=True)
df['birthdate_clean'] = pd.to_datetime(df['birthdate_clean'], dayfirst=False)
df['age'] = (pd.to_datetime('today') - df['birthdate_clean']).dt.days // 365.25
```

```python
df['efficiency'] = df['avgPoints'] + df['avgRebounds'] + df['avgAssists'] - df['avgTurnovers']
df['efficiency_posi'] = df.groupby('position')['efficiency'].transform(lambda x: (x - x.mean())/ x.std())
```

```python
df['status'].unique()
df["status"] = df["status"].map({"active": 1, "inactive": 0})
```

```python
df_ = df.drop(['name','positionId','playerId','playerSlug','college','birthplace','draftinfo','htwt','rankingSalary','team','birthdate','birthdate_clean','experience'],axis=1)
```

```python
df_.head(3)
```

```python
import numpy as np
import pandas as pd

from sklearn.model_selection import train_test_split
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
import sklearn

def rmse(y_true, y_pred):
    return float(np.sqrt(mean_squared_error(y_true, y_pred)))

print("sklearn:", sklearn.__version__)
```

## Definição amostragem

```python
# Ajuste aqui se o alvo não for 'salary'
target_col = "salary"

# Garante que df_ existe
assert target_col in df_.columns, "Coluna alvo 'salary' não encontrada em df_"

X = df_.drop(columns=[target_col])
y = df_[target_col].copy()

# Split treino/teste
X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.20,
    random_state=43,  # (42 + 1 como você usou)
    shuffle=True
)

X_train.shape, X_test.shape, y_train.shape, y_test.shape
```

## Pipelines de Transformação

```python
# Detecta dtypes após o split (importante!)
num_cols = X_train.select_dtypes(include=["number"]).columns.tolist()
cat_cols = X_train.select_dtypes(include=["object", "string"]).columns.tolist()

# Imputers e scalers
num_pipeline = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler())
])

# Compatibilidade de versão do OneHotEncoder
if sklearn.__version__ >= "1.2":
    ohe = OneHotEncoder(sparse_output=False, handle_unknown="ignore")
else:
    ohe = OneHotEncoder(sparse=False, handle_unknown="ignore")

cat_pipeline = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="constant", fill_value="missing")),
    ("encoder", ohe)
])

full_pipeline = ColumnTransformer(transformers=[
    ("num", num_pipeline, num_cols),
    ("cat", cat_pipeline, cat_cols)
])

# Fit no TREINO e transform nos dois
X_train_prepared = full_pipeline.fit_transform(X_train)
X_test_prepared  = full_pipeline.transform(X_test)

X_train_prepared.shape, X_test_prepared.shape
```

## Regressão Linear

```python
lin_reg = LinearRegression()
lin_reg.fit(X_train_prepared, y_train)

y_pred_train = lin_reg.predict(X_train_prepared)
y_pred_test  = lin_reg.predict(X_test_prepared)

rmse_train  = rmse(y_train, y_pred_train)
mae_train   = mean_absolute_error(y_train, y_pred_train)
r2_train    = r2_score(y_train, y_pred_train)

rmse_test   = rmse(y_test, y_pred_test)
mae_test    = mean_absolute_error(y_test, y_pred_test)
r2_test    = r2_score(y_test, y_pred_test)

print("LinearRegression")
print(f"RMSE treino: {rmse_train:,.0f} | MAE treino: {mae_train:,.0f} | R² treino: {r2_train:,.3f}")
print(f"RMSE teste : {rmse_test:,.0f} | MAE teste : {mae_test:,.0f} | R² teste: {r2_test:,.3f}")
```

# RandomForestRegressor

```python
from sklearn.ensemble import RandomForestRegressor

rf = RandomForestRegressor(
    n_estimators=300,
    max_depth=None,
    n_jobs=-1,
    random_state=43
)

rf.fit(X_train_prepared, y_train)

rf_pred_train = rf.predict(X_train_prepared)
rf_pred_test = rf.predict(X_test_prepared)

rf_rmse_train  = rmse(y_train, rf_pred_train)
rf_mae_train   = mean_absolute_error(y_train, rf_pred_train)
rf_r2_train    = r2_score(y_train, rf_pred_train)

rf_rmse_test = rmse(y_test, rf_pred_test)
rf_mae_test  = mean_absolute_error(y_test, rf_pred_test)
rf_r2_test  = r2_score(y_test, rf_pred_test)

print("RandomForest")
print(f"RMSE treino: {rf_rmse_train:,.0f} | MAE treino: {rf_mae_train:,.0f} | R² treino: {rf_r2_train:,.3f}")
print(f"RMSE teste: {rf_rmse_test:,.0f} | MAE teste: {rf_mae_test:,.0f} | R² teste: {rf_r2_test:,.3f}")

```

## RandomForestRegressor ( Hiperparametros )

```python
from sklearn.ensemble import RandomForestRegressor
from sklearn.model_selection import RandomizedSearchCV
from scipy.stats import randint

# Definição do modelo base
rf = RandomForestRegressor(random_state=43, n_jobs=-1)

# Espaço de busca dos hiperparâmetros
param_dist = {
    "n_estimators": randint(200, 1000),        # nº de árvores
    "max_depth": randint(5, 30),               # profundidade máxima
    "min_samples_split": randint(2, 20),       # min. amostras p/ split
    "min_samples_leaf": randint(1, 10),        # min. amostras por folha
    "max_features": ["sqrt", "log2", 0.3, 0.5] # nº de features p/ split
}

# Configuração do RandomizedSearchCV
random_search = RandomizedSearchCV(
    rf,
    param_distributions=param_dist,
    n_iter=30,              # nº de combinações testadas (aumente p/ mais precisão)
    scoring="neg_root_mean_squared_error", # métrica = RMSE
    cv=3,                   # cross-validation com 3 folds
    random_state=43,
    n_jobs=-1,
    verbose=2
)

# Treinamento no dataset preparado
random_search.fit(X_train_prepared, y_train)

# Melhor combinação de parâmetros
print("Melhores parâmetros:", random_search.best_params_)

# Melhor estimador treinado
best_rf = random_search.best_estimator_

# Avaliação em treino e teste
rf_train_pred = best_rf.predict(X_train_prepared)
rf_test_pred  = best_rf.predict(X_test_prepared)

rf_rmse_train = rmse(y_train, rf_train_pred)
rf_mae_train   = mean_absolute_error(y_train, rf_train_pred)
rf_r2_train    = r2_score(y_train, rf_train_pred)

rf_rmse_test  = rmse(y_test, rf_test_pred)
rf_mae_test  = mean_absolute_error(y_test, rf_test_pred)
rf_r2_test  = r2_score(y_test, rf_test_pred)

print("RandomForest otimizado")
print(f"RMSE treino: {rf_rmse_train:,.0f} | MAE treino: {rf_mae_train:,.0f} | R² treino: {rf_r2_train:,.3f}")
print(f"RMSE teste: {rf_rmse_test:,.0f} | MAE teste: {rf_mae_test:,.0f} | R² teste: {rf_r2_test:,.3f}")

```

```python
import pandas as pd
import matplotlib.pyplot as plt

# --- PASSO 1: Preparar os Dados ---
# Supondo que 'df' seja seu DataFrame com os dados dos jogadores

# Crie a coluna que identifica os salários "superfaturados"
df['superfaturado'] = df['salario_real'] > (df['salario_previsto'] + df['margem_de_erro'])

# Conte quantos jogadores estão em cada categoria (True para superfaturado, False para não)
contagem_status = df['superfaturado'].value_counts()

# Mapeia os valores True/False para rótulos mais claros
contagem_status.index = contagem_status.index.map({True: 'Salários Superfaturados', False: 'Salários Dentro do Esperado'})


# --- PASSO 2: Gerar o Gráfico de Rosca (Donut Chart) ---
fig, ax = plt.subplots(figsize=(8, 8))

# Cores para o gráfico
cores = ['#e10157', '#dddddd'] # Cor de destaque para os superfaturados, cinza para o resto

# Plota o gráfico de pizza
wedges, texts, autotexts = ax.pie(
    contagem_status,
    labels=contagem_status.index,
    autopct='%1.1f%%', # Formata o percentual com 1 casa decimal
    startangle=90,
    colors=cores,
    pctdistance=0.85, # Distância do percentual em relação ao centro
    textprops={'fontsize': 12, 'fontweight': 'bold'}
)

# Transforma a pizza em uma rosca, desenhando um círculo branco no centro
centre_circle = plt.Circle((0,0),0.70,fc='white')
fig.gca().add_artist(centre_circle)

# Ajustes de estilo nos textos
plt.setp(autotexts, size=14, weight="bold", color="white")
plt.setp(texts, size=12)

# Título
ax.set_title("Percentual de Jogadores com Salários Superfaturados", fontsize=16, fontweight='bold')

ax.axis('equal')  # Garante que a pizza seja um círculo.
plt.tight_layout()
plt.show()
```
